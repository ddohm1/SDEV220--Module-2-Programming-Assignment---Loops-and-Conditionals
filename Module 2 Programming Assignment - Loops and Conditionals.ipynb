```python
small = True
green = False

if small and green:
    print("Pea")
elif not small and green:
    print("Watermelon")
elif small and not green:
    print("Cherry")
elif not small and not green:
    print("Pumpkin")
else:
    print("Uh oh!")
```

```python
guess_me = 7
number = 1
while True:
    if number < guess_me:
        print("Too low")
    elif number == guess_me:
        print("Found it!")
        break
    elif number > guess_me:
        print("oops")
        break
    number += 1
```
